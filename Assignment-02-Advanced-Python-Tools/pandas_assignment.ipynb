Objective Learn to create and manipulate dataframes for data analysis.


```python
import pandas as pd
```

The Pandas library is imported using the alias `pd`.

 Pandas is used for data manipulation and analysis with structures such as DataFrames.

**1. Create a DataFrame with the following data:**

```python
data = {
 'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
 'Age': [24, 27, 22, 32, 29],
 'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
 'Salary': [45000, 54000, 50000, 62000, 47000]
 }
data
```

```python
df = pd.DataFrame(data)
print(df)
```

DataFrame stores data in a tabular format with rows and columns, making it easy to analyze and manipulate data.

a. Print the first five rows of the DataFrame.

```python
df.head(5)
```

b. Get the summary statistics of the 'Age' and 'Salary' columns.


```python
print(df[['Age', 'Salary']].describe())
```

selected the specific columns and then using .describe() we got every details such as count, mean, minimum, maximum, and standard deviation.

c. Calculate the average salary of employees in the 'HR' department.

```python
print(df[df['Department']=="HR"]['Salary'].mean())
```

The employees belonging to the HR department are filtered at df[df['Department']=="HR"] , and their average salary is calculated using the `mean()` function.

2. Add a new column, 'Bonus', which is 10% of the salary.

```python
df['Bonus'] = df['Salary'] * 0.10
df
```

In this step, a new column named 'Bonus' is added to the DataFrame. The bonus is calculated as 10% of each employee’s salary. using mathemetical operation *df['Salary'] * 0.10*

3. Filter the DataFrame to show employees aged between 25 and 30.


```python
print(df[(df['Age'] >= 25) & (df['Age'] <= 30)])
```

The DataFrame is filtered to display employees whose age is between 25 and 30. with *df[(condition1) & (condition2)]*

4. Group the data by 'Department' and calculate the average salary for each
department.

```python
print(df.groupby('Department')['Salary'].mean())
```

The data is grouped by 'Department' using the *df.groupby()* and then average salary is count by *['salary'].mean()* where after df.groupby()['salary'] is added which is done to select the salary column

5. Sort the DataFrame by 'Salary' in ascending order and save the result to a new CSV
file.


```python
sorted_df = df.sort_values(by='Salary')
sorted_df
sorted_df.to_csv('sortedby_salary.csv', index=False)
```

The dataframe is sorted using *df.sort_values(by='Column_name')*

then it is saved to the new csv file via *.to_csv("New_csv_name")*
