# **Assignment 3 - Working with Matplotlib**

Matplotlib is imported to create different types of graphs and charts for data visualization.

```python
import matplotlib.pyplot as plt
import pandas as pd
```

1. Create a simple line plot for the following data:

 a. Plot the data.
 b. Customize the plot by adding a title, axis labels, and a grid.

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]

plt.plot(x,y)

plt.title("Line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid(True)

plt.show()
plt.show()
```

*plt.plot(x, y)* - Creates line graph.

*plt.title()* - Adds graph title.

*xlabel() and ylabel()* - Adds axis labels.

*grid(True)* - Shows background grid.

*show()* - Displays graph.

Without *show()*, graph may not appear properly.

2. Create a bar graph to represent the marks scored by students in a subject:

a. Plot the data as a bar graph.
 b. Customize the colors and add a title.

```python
students = ['John', 'Jane', 'Alice', 'Bob']
marks =  [75, 85, 60, 90]
```

```python
plt.bar(students,marks,color=['blue','green','red','violet'])

plt.title("Bar graph")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

***plt.bar(X-axis, Y-axis)*** - To create Bar graph of X with relation to Y

***color=[]*** parameter is used to define colors of each elements of X

3. Create a pie chart to represent the percentage distribution of a company’s revenue
from different regions:

a. Create a pie chart with the region names as labels.
 b. Highlight the region with the highest revenue.

```python
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]

explode = [0.1, 0, 0, 0] #explode is used to assign highlight value to the element

plt.pie(revenue, labels=regions, explode=explode) #passing the explode variable value to explode parameter

plt.title("Pie chart")

plt.show()
```

**`plt.pie(revenue)`** - to create pie chart of revenue

*`labels=regions`* - Adds names of regions around pie chart.

*`explode`* - Highlights one section.

Here: 0.1

slightly separates highest revenue region.

`autopct` - Displays percentages.

4. Generate a histogram to show the frequency distribution of randomly generated
integers between 1 and 100 (sample size = 1000).

Breaking down :
1. randomly generate integers between 1 to 100
2.  Generate histogram to show the frequency distribution of them

```python
import numpy as np

data = np.random.randint(1, 101, 1000)
```

`np.random.randint(1, 101, 1000)`

Creates:

random integers

from 1 to 100

total 1000 values

```python
plt.hist(data,bins=10)

plt.title("Histogram of Random Numbers")
plt.xlabel("Number Range")
plt.ylabel("Frequency")

plt.show()
```

**`plt.hist(data)`** - Creates histogram.

`bins=10`
Divides data into 10 groups/ranges.
