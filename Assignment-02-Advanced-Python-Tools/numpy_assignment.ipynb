1. Create a 1D NumPy array with integers from 1 to 20.

```python
arr=np.arange(1,21)
print(arr)
```

a one-dimensional NumPy array containing integers from 1 to 20 is created using the np.arange() function. NumPy arrays are faster and more efficient for numerical operations compared to normal Python lists.

Perform the following operations:

a. Calculate the sum, mean, median, and standard deviation of the elements in the array.

```python
print(np.sum(arr))
#OR
arr.sum()
```

```python
print(np.mean(arr))
#OR
arr.mean()
```

```python
print(np.median(arr))
```

```python
print(np.std(arr))
#OR
arr.std()
```

numpy has builtin mathemetical functions which can be used to perform various mathematical operations as shown.

b. Find the indices of elements greater than 10 in the array.

```python
i = np.where(arr>10)
print(i)
```

np.where(condition) helps in finding elements of array with specific conditions

**2. Create a 2D NumPy array of shape 4 X 4 with numbers ranging from 1 to 16.**

a. Print the array.


```python
arr = np.arange(1,17).reshape(4,4)
print(arr)
```

To create any Dimensional array we can use np.arange(start,end).reshape(shape) to create and nD type of array.

where reshape helps in shaping the array

b. Find the transpose of the array.

```python
np.transpose(arr)
#OR
arr.T
```

For transpose of a matrix or array we can use name.T or np.transpose(name)

c. Calculate the row-wise and column-wise sums of the array.

```python
print("Row wise sum:" ,np.sum(arr,axis=1))
print("Column wise sum:" , np.sum(arr,axis=0))
```

we use axis parameter to determine the operation , axis=1 decides ROW wise operations , axis=90 decides COLUMN wise operations
